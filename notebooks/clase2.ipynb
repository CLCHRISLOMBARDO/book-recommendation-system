```python
import sqlite3
import pandas as pd
```

```python
DATA_BASE = "dataset/data.db"
query = """ select id_lector 
            from interacciones 
            GROUP BY id_lector ORDER BY COUNT(*)
            HAVING count(*)>= 20
            ORDER BY COUNT(*) DESC limit 20
            """
con = sqlite3.connect(DATA_BASE)
df=pd.read_sql_query(query , con)
con.close()
```
