Entonces, el plan de pasar el id del cliente no me parece bueno. Mas que pasar su dni, hay que pasar su adn. En vez de quien es, pasarle su perfil.

Ideas de feat eng : 
- Variables de comportamiento : 
    - rating_promedio_usuario : ¿Es un crítico exigente que suele poner 4, o es generoso y pone 9?
    - volumen_lectura : Cantidad total de libros calificados (esto ayuda al modelo a tratar diferente a un lector hardcore que a uno de un solo libro)
    - desviacion_estandar_rating: ¿Sus notas varían mucho o siempre pone lo mismo?
- Variables de afinidad
    - genero_favorito_usuario: El género de libro que más lee.
    - ratio_libros_antiguos_vs_nuevos: Si tiende a leer clásicos o novedades (usando anio_edicion).
Nota para el caso de 1 solo libro: Para los lectores con una sola interacción, su rating_promedio será exactamente la nota de ese único libro. Esto puede causar leakage (fuga de información) durante el entrenamiento. Se suele calcular el promedio excluyendo la fila actual, o rellenando con el promedio global de la base.

- Target encoding
Si de todas formas quieres que el modelo use explícitamente el ID para capturar algún efecto específico de ese usuario que las otras variables no explican, no puedes usar One-Hot Encoding (tendrías miles de columnas vacías) ni Label Encoding (el árbol interpretará que el lector 100 vale el doble que el 50).

Puedes usar Target Encoding: reemplazar el id_lector por la media del rating que ese lector suele dar.
Para resolver tu preocupación sobre los usuarios con un solo libro, es obligatorio aplicar una fórmula de suavizado. El suavizado tira el promedio del usuario hacia el promedio global de todos los lectores si el usuario tiene muy pocos datos.
